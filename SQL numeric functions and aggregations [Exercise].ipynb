### Exercise 1

Display the average unit price of products in the 'Products' table.

```python
%%sql
SELECT
    AVG(UnitPrice) AS Average_Unit_Price
FROM
    Products;
```

### Exercise 2


Show the total quantity of products sold for each ProductID. Use the 'OrderDetails' table.

```python
%%sql
SELECT
    ProductID,
    SUM(ProductID) AS Total_Quantity_Sold
FROM
    "Order Details"
GROUP BY
    ProductID;
```

### Exercise 3


Find the maximum, minimum, and average freight from the 'Orders' table.

```python
%%sql
SELECT
    MIN(Freight) AS min_freight,
    MAX(Freight) AS max_freight,
    AVG(Freight) AS avg_freight
FROM
    Orders;
```

### Exercise 4

How many unique customers have made orders? Use the 'Orders' table.

```python
%%sql
SELECT
    COUNT(DISTINCT CustomerID) AS Number_0f_customers
FROM
    Orders;
```

## Solutions

### Exercise 1

```python
%%sql

SELECT 
    AVG(UnitPrice) 
FROM 
    Products;
```

### Exercise 2

```python
%%sql

SELECT 
    ProductID,
    SUM(Quantity) as TotalQuantity
FROM 
    OrderDetails
GROUP BY 
    ProductID;
```

### Exercise 3

```python
%%sql

SELECT 
    MAX(Freight) as MaxFreight,
    MIN(Freight) as MinFreight,
    AVG(Freight) as AvgFreight
FROM 
    Orders;
```

### Exercise 4

```python
%%sql

SELECT 
    COUNT(DISTINCT CustomerID) 
FROM 
    Orders;
```

### Exercise 5

```python
%%sql

SELECT 
    OrderID, 
    SUM(UnitPrice * Quantity * (1 - Discount)) as TotalPrice 
FROM 
    OrderDetails
GROUP BY 
    OrderID;
```

### Challenge question

```python
%%sql

SELECT 
    OrderID, 
    SUM(Quantity) as TotalQuantity 
FROM 
    OrderDetails
GROUP BY 
    OrderID
ORDER BY 
    TotalQuantity DESC 
LIMIT 5;
```
