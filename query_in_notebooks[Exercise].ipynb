# Exercise 1
#### Find the names of all customers from Germany in the Northwind database.

```python
%%sql
SELECT
    ContactName
FROM
    customers
WHERE
    Country = "Germany";
```

# Exercise 2
### Find all products in the Northwind database that have a unit price greater than 30 and are discontinued.

### Note: The product has been discontinued when the column Discontinued is equal to 1.

```python
%%sql
SELECT
    ProductName
FROM
    products
WHERE 
    UnitPrice > 30
AND
   Discontinued = 1;
```

## Exercise 3
#### Find the contact names and phone numbers of customers from either 'USA' or 'Canada'.

```python
%%sql
SELECT
    ContactName,Phone
FROM
    customers
WHERE
    Country = "USA"
OR
    Country = "Canada";
```

# Exercise 4
#### From the 'Orders' table, select all orders that were placed by 'VINET', 'QUICK' or 'SUPRD' and were shipped via ### shipper 1 or 2.

```python
%%sql
SELECT
    *
FROM
    orders
WHERE
    (CustomerID = "VINET"
OR 
    CustomerID = "QUICK"
OR
    CustomerID = "SUPRD")
AND
    (ShipVia = 1
OR 
    ShipVia = 2);
```

# Challenge question
### Find the employees in the Northwind database who are either Sales Representatives or Sales Managers and are 
#### based in London.

```python
%%sql
SELECT
    FirstName,
    LastName
FROM
    Employees
WHERE
    (Title = 'Sales Representative' OR Title = 'Sales Manager')
    AND City = 'London';
```

### Exercise 4  using IN

```python
%%sql
SELECT
    *
FROM
    orders
WHERE
    CustomerID IN ("VINET","QUICK","SUPRD")
AND
    ShipVia IN (1,2);
```
